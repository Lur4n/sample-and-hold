### Seja um sinal analógico dado por:
> $v(t) = 1 + cos(2πt) V.$

Para

>$ 0 ≤ t < 1 $ (*em segundos*)

---

Plot do gráfico de sinal após o processo de *sample and hold* para um conjunto de $20$ amostras, igualmente espaçadas e contadas a partir de $t = 0$.


#### Parametrização

```python
import matplotlib.pyplot as plt
import numpy as np

n_amostras = 20
x = np.linspace(0, 1, n_amostras)
y = 1 + np.cos(2*np.pi*x) 
dpi_graficos = 100
```

#### Amostra

```python
def plot_sinal_analogico(x, y):
  # 1. Criação do Objeto Figure e Axes
  fig, ax = plt.subplots(figsize=(10, 5), dpi=dpi_graficos)
  # 2. Configuração do Plot da Curva
  ax.plot(x, y, color="#007ACC", linewidth=1, linestyle="-", label="f(x) manual")
  # 3. Títulos e Rótulos dos Eixos
  ax.set_xticks(x)
  ax.set_title("Sinal Analógico Contínuo (Entrada do ADC)", fontsize=12, fontweight="bold")
  ax.set_xlabel("Tempo (segundos)", fontsize=10)
  ax.set_ylabel("Tensão (Volt)", fontsize=10)
  # 4. Grade e Destaque nos Eixos Zero (Origem)
  ax.grid(True, which="both", linestyle="--", alpha=0.5)
  ax.axhline(0, color="red", linewidth=0.8)
  # 5. Limites e Ticks dos Eixos
  ax.set_xlim(min(x), max(x))
  ax.set_ylim(min(y)-0.18, max(y)+0.18)
  ax.legend(loc="upper right", frameon=True, facecolor="white", edgecolor="none")
  
  plt.tight_layout()
  plt.show()
```

#### Trem de Pulsos

```python
def trem_de_pulsos(x, y):
  # 1. Criação do Objeto Figure e Axes
  fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 5), dpi=dpi_graficos)

  # Trem de pulsos unitários puro p(t)
  ax1.stem(x, y, linefmt="-", markerfmt="go", label="Pulsos Unitários")
  ax1.set_title("1. Trem de Pulsos Unitários p(t) (Trem de Impulsos de Amostragem)", fontweight="bold")
  ax1.set_xticks(x)
  ax1.set_xlabel("Tempo (segundos)")
  ax1.set_xlim(min(x), max(x))
  ax1.set_ylabel("Amplitude")
  ax1.set_ylim(min(y)-0.18, max(y)+0.18)
  ax1.grid(True, linestyle="--", alpha=0.5)
  ax1.legend(loc="upper right")

  # Sinal continuo x Amostras moduladas (PAM)
  # 2. Configuração do Plot da Curva
  ax2.plot(x, y, color="#007ACC", alpha=0.4, linewidth=1, label="Sinal 'Contínuo'")
  ax2.stem(x, y, linefmt="-", markerfmt="go", label="Amostras")
  ax2.set_title("2. Sinal Amostrado (Senoide × Trem de Pulsos)", fontweight="bold")
  ax2.set_xticks(x)
  ax2.set_xlabel("Tempo (segundos)")
  ax2.set_xlim(min(x), max(x))
  ax2.set_ylabel("Tensão (Volt)")
  ax2.set_ylim(min(y)-0.18, max(y)+0.18)
  ax2.grid(True, linestyle="--", alpha=0.5)
  ax2.legend(loc="upper right")

  plt.tight_layout()
  plt.show()
```

#### Sample and Hold

```python
def sample_and_hold(x, y):
  fig, ax3 = plt.subplots(figsize=(10, 5), dpi=dpi_graficos)

  # Retenção
  ax3.step(x , y, where="post", label="Sinal Retido", color="#007ACC", linewidth=1)
  ax3.set_title("Sinal Amostrado e Retido (Sample and Hold)", fontweight="bold")
  ax3.set_xticks(x)
  ax3.set_xlabel("Tempo (s)")
  ax3.set_xlim(min(x), max(x))
  ax3.set_ylabel("Tensão (Volt)")
  ax3.set_ylim(min(y)-0.18, max(y)+0.18)
  ax3.grid(True, linestyle="--", alpha=0.5)
  ax3.axhline(0, color="red", linewidth=0.8)
  ax3.legend(loc="upper right")

  plt.tight_layout()
  plt.show()
```

#### Plotagens

```python
plot_sinal_analogico(x, y)
trem_de_pulsos(x, y)
sample_and_hold(x, y)
```
